## Part 1 — Understanding Local AI

**Cloud AI**
```
Python App → Internet → Cloud API → LLM
```
**Local AI**
```
Python App → localhost:11434 → Ollama → Local LLM → GPU / CPU
```
Advantages to investigate: **privacy**, **offline operation**, **no per-token cost**.
Trade-off: local models may be **less capable or slower** than frontier cloud models, depending on hardware and model size.

### Exercise 4 — Run the model
In the terminal you would use `ollama run qwen3.5`. Here we do the same thing through the API with a helper function.

````python
import time, json

def generate(prompt, model=None, system=None, fmt=None, images=None, options=None, think=None):
    """Call Ollama /api/generate (non-streaming) and return the full JSON dict
    plus the measured wall-clock time under the key 'wall_s'."""
    payload = {"model": model or MODEL, "prompt": prompt, "stream": False}
    if system:  payload["system"] = system
    if fmt:     payload["format"] = fmt          # "json" or a JSON schema
    if images:  payload["images"] = images       # list of base64 strings
    if options: payload["options"] = options
    if think is not None: payload["think"] = think  # False disables 'thinking' on reasoning models
    t0 = time.perf_counter()
    r = requests.post(f"{OLLAMA}/api/generate", json=payload, timeout=600)
    r.raise_for_status()
    out = r.json()
    out["wall_s"] = time.perf_counter() - t0
    return out

def ask(prompt, **kw):
    """Shortcut: return only the text, and print timing."""
    out = generate(prompt, **kw)
    print(f"[{out['wall_s']:.1f}s]")
    return out["response"]
````

````python
for p in [
    "Explain what artificial intelligence is in five sentences.",
    "Explain artificial intelligence to a 10-year-old child.",
    "Explain artificial intelligence to a computer science professor.",
]:
    print("="*80, "\nPROMPT:", p, "\n")
    print(ask(p))
````

````python
import math
from typing import Optional

def is_prime_qwen(n: int) -> bool:
    """
    Determines whether a given integer 'n' is a prime number.

    A prime number is a natural number greater than 1 that has no positive 
    divisors other than 1 and itself.

    Args:
        n (int): The integer number to check for primality.

    Returns:
        bool: True if 'n' is a prime number, False otherwise.
    
    Raises:
        TypeError: If the input 'n' is not an integer.
    """
    
    # 1. Handle base cases (numbers less than or equal to 1 are not prime)
    if n <= 1:
        return False
    
    # 2. Handle the number 2 separately (it is the only even prime number)
    if n == 2:
        return True

    # 3. Eliminate other even numbers quickly
    if n % 2 == 0:
        return False

    # 4. Check for factors from 3 up to the square root of n
    # We increment by 2 to skip even numbers (3, 5, 7, etc.)
    limit = int(math.isqrt(n))
    for i in range(3, limit + 1, 2):
        if n % i == 0:
            return False
            
    return True

def is_prime_gemma(number: int) -> bool:
    """
    Determines whether a given number is a prime number.

    Algorithm:
    The function implements the trial division method for primality testing.
    It iterates from 2 up to the square root of the input number.
    For each number in this range, it checks if the input number is divisible by the current iterator.
    If the number is divisible by any iterator, it means the number is not prime, and the function returns False.
    If the loop completes without finding any divisors, the number is prime, and the function returns True.

    We only need to check divisibility up to the square root of the number because
    if a number has a divisor greater than its square root, it must also have a divisor smaller than its square root.

    Args:
        number: An integer to be checked for primality.

    Returns:
        True if the number is prime, False otherwise.
    """

    # Handle trivial cases:
    if number <= 1:
        return False  # Numbers less than or equal to 1 are not prime.
    if number <= 3:
        return True  # 2 and 3 are prime.

    # If the number is divisible by 2 or 3, it's not prime:
    if number % 2 == 0 or number % 3 == 0:
        return False

    # Iterate from 5 up to the square root of the number, incrementing by 6.
    # This optimization is based on the fact that all primes greater than 3 can be expressed in the form 6k ± 1.
    i = 5
    while i * i <= number:
        if number % i == 0 or number % (i + 2) == 0:
            return False
        i += 6

    return True
````

````python
def is_prime_reference(n: int) -> bool:
    if n < 2: return False
    if n < 4: return True
    if n % 2 == 0: return False
    i = 3
    while i * i <= n:
        if n % i == 0: return False
        i += 2
    return True

# Paste the model's function below as `candidate` and compare:
candidates = [is_prime_qwen, is_prime_gemma]
for candidate in candidates:
    assert all(candidate(n) == is_prime_reference(n) for n in range(-5, 2000)), "Incorrect!"
    print(f"✅ {candidate.__name__} candidate matches the reference on 0..2000")
````

## Part 9 — Add Conversation Memory
The previous app treats each question independently. We now send the **conversation history** with each request.

````python
history = []

def ask_ai_memory(question):
    history.append(f"User: {question}")
    prompt = "\n".join(history) + "\nAssistant:"
    data = {"model": MODEL, "prompt": prompt, "stream": False}
    response = requests.post(URL, json=data)
    response.raise_for_status()
    answer = response.json()["response"]
    history.append(f"Assistant: {answer}")
    return answer

def chat_loop_memory():
    while True:
        question = input("\nYou: ")
        if question.lower() == "exit":
            break
        print("\nAI:", ask_ai_memory(question))

# chat_loop_memory()   # <- uncomment for interactive use
````

#### Better approach: the `/api/chat` endpoint
Ollama also has a chat endpoint taking structured `messages` (with `system` / `user` / `assistant` roles) and applying the model's own chat template.

````python
def chat(messages, model=None, **kw):
    payload = {"model": model or MODEL, "messages": messages, "stream": False, **kw}
    r = requests.post(f"{OLLAMA}/api/chat", json=payload, timeout=600)
    r.raise_for_status()
    return r.json()

msgs = [{"role": "system", "content": "You are a concise assistant."},
        {"role": "user", "content": "My name is Ahmed."}]
reply = chat(msgs)["message"]
msgs.append(reply)
msgs.append({"role": "user", "content": "What is my name?"})
print(chat(msgs)["message"]["content"])
````

## Part 11 — Structured Output
### Exercise 11
First, the naive approach: ask for "ONLY JSON" in the prompt.

````python
json_prompt = """Analyze the following student:

Name: Ahmed
Age: 21
Grade: 15.5
Absences: 2

Return ONLY JSON with this structure:

{
  "name": "...",
  "age": 0,
  "grade": 0,
  "risk": "...",
  "recommendation": "..."
}"""

raw = ask(json_prompt, think=False)
print(raw)
````

````python
import re

def extract_json(text):
    """Robust parsing: strip ```json fences and grab the first {...} block."""
    text = re.sub(r"```(?:json)?", "", text).strip()
    m = re.search(r"\{.*\}", text, re.DOTALL)
    return json.loads(m.group(0))

try:
    print(extract_json(raw))
except Exception as e:
    print("❌ Could not parse:", e)
````

Better: use Ollama's **structured outputs** — pass a JSON Schema in `format` so the output is constrained to be valid.

````python
schema = {
    "type": "object",
    "properties": {
        "name": {"type": "string"},
        "age": {"type": "integer"},
        "grade": {"type": "number"},
        "risk": {"type": "string", "enum": ["low", "medium", "high"]},
        "recommendation": {"type": "string"},
    },
    "required": ["name", "age", "grade", "risk", "recommendation"],
}

out = generate(json_prompt, fmt=schema, think=False)
student = json.loads(out["response"])
print(json.dumps(student, indent=2))
print("Type of grade:", type(student["grade"]))
````

### Question
Why is structured output important when integrating an LLM into a software application?

**✍️ Your answer:**

Structured output allows a non deterministic schema to to conform to a structure, which is very important when designing software applications, the reason being is that otherwise any attempt to parse what the model is outputting would be just fragile heuristics, json output allows us to enforce a structure which makes all output conform the shape the app / API expects

## Part 13 — Mini-RAG
We create four documents, then build a **local retriever** that picks the most relevant one for a question and feeds it to the LLM.

````python
docs = {
"ai.txt": """Artificial intelligence (AI) is the field of computer science that builds systems able to perform tasks normally requiring human intelligence, such as reasoning, perception, language understanding and decision making. AI includes machine learning, expert systems, robotics and natural language processing.""",

"machine_learning.txt": """Machine learning is a branch of artificial intelligence. Supervised learning uses labelled data to learn a mapping from inputs to outputs (classification, regression). Unsupervised learning works with unlabelled data to discover structure such as clusters. Overfitting happens when a model memorises training data and generalises poorly.""",

"reinforcement_learning.txt": """Reinforcement learning learns through interaction with an environment using rewards and penalties. An agent observes a state, chooses an action, and receives a reward. The goal is to learn a policy that maximises the cumulative reward. Q-learning and policy gradients are common algorithms.""",

"deep_learning.txt": """Deep learning uses artificial neural networks with many layers to learn representations from large amounts of data. Convolutional neural networks are used for images, recurrent networks and transformers for sequences and language. Training deep networks usually requires GPUs.""",
}

for name, text in docs.items():
    with open(f"documents/{name}", "w", encoding="utf-8") as f:
        f.write(text)
os.listdir("documents")
````

````python
%%writefile rag.py
"""Mini-RAG: a tiny local retriever + prompt builder (keyword TF-IDF, no extra libraries)."""
import math, os, re
from collections import Counter

STOPWORDS = set("""a an the of to in on and or is are was were be by for with as at from that this
it its what which who how why when where do does did can could should would about into using use
learn learns through between such""".split())


def tokenize(text):
    return [w for w in re.findall(r"[a-zA-Z]+", text.lower()) if w not in STOPWORDS]


def load_documents(folder="documents"):
    docs = {}
    for fn in sorted(os.listdir(folder)):
        if fn.endswith(".txt"):
            with open(os.path.join(folder, fn), encoding="utf-8") as f:
                docs[fn] = f.read()
    return docs


class Retriever:
    def __init__(self, docs):
        self.docs = docs
        self.tokens = {n: Counter(tokenize(t)) for n, t in docs.items()}
        df = Counter(w for c in self.tokens.values() for w in c)
        N = len(docs)
        self.idf = {w: math.log((N + 1) / (d + 0.5)) for w, d in df.items()}

    def score(self, question):
        q = tokenize(question)
        return {
            name: sum(self.idf.get(w, 0) * (1 + math.log(c[w])) for w in set(q) if c[w] > 0)
            for name, c in self.tokens.items()
        }

    def top(self, question, k=1, min_score=0.01):
        ranked = sorted(self.score(question).items(), key=lambda x: -x[1])
        return [(n, s) for n, s in ranked[:k] if s >= min_score]

    def context(self, question, k=1):
        hits = self.top(question, k)
        return hits, "\n\n".join(f"[{n}]\n{self.docs[n]}" for n, _ in hits)


def build_prompt(question, context):
    return f"""Answer the question using only the CONTEXT below.
If the answer is not contained in the context, say: "Information not available."

CONTEXT:
{context if context else '(no relevant document found)'}

QUESTION:
{question}
"""
````

````python
import importlib, rag
importlib.reload(rag)

retriever = rag.Retriever(rag.load_documents("documents"))

def rag_answer(question, k=1, model=None):
    hits, ctx = retriever.context(question, k)
    out = generate(rag.build_prompt(question, ctx), model=model, think=False)
    return hits, out["response"], out["wall_s"]

for q in ["What is reinforcement learning?",
          "Which hardware is needed to train neural networks with many layers?",
          "What is supervised learning?",
          "Who won the 2018 World Cup?"]:
    hits, ans, t = rag_answer(q)
    print(f"Q: {q}\n  retrieved: {hits}\n  A: {ans.strip()}\n  ({t:.1f}s)\n")
````

**Architecture**
```
Documents → Local Retriever → Relevant Context → Ollama → Local LLM → Answer
```
This is the fundamental idea behind **Retrieval-Augmented Generation (RAG)**.

#### Optional extension — semantic retrieval with embeddings
Keyword matching fails when the question uses different words than the document. Ollama can also compute embeddings locally (`ollama pull nomic-embed-text`).

````python
EMBED_MODEL = "nomic-embed-text"
!ollama pull {EMBED_MODEL}
````

````python
import numpy as np

def embed(texts):
    r = requests.post(f"{OLLAMA}/api/embed", json={"model": EMBED_MODEL, "input": texts})
    r.raise_for_status()
    return np.array(r.json()["embeddings"])

def semantic_top(question, docs, k=1):
    names = list(docs)
    D = embed([docs[n] for n in names]); q = embed([question])[0]
    sims = D @ q / (np.linalg.norm(D, axis=1) * np.linalg.norm(q))
    return sorted(zip(names, sims), key=lambda x: -x[1])[:k]

print(semantic_top("What hardware accelerates network training?", rag.load_documents("documents"), k=2))
````

## Part 16 — Performance Benchmark
We measure: response time, model size, tokens generated, tokens/second, RAM and GPU consumption.

Ollama returns precise counters: `eval_count` (generated tokens), `eval_duration` (ns), `prompt_eval_count`, `load_duration`.

````python
!pip install -q psutil
````

````python
%%writefile benchmark.py
"""Benchmark local Ollama models on the same prompt."""
import time, requests, psutil, pandas as pd

OLLAMA = "http://localhost:11434"
PROMPT = "Write a 500-word explanation of reinforcement learning."


def unload(model):
    requests.post(f"{OLLAMA}/api/generate", json={"model": model, "keep_alive": 0})
    time.sleep(2)


def loaded_info(model):
    for m in requests.get(f"{OLLAMA}/api/ps").json()["models"]:
        if m["name"].startswith(model.split(":")[0]):
            return m["size"] / 1e9, m.get("size_vram", 0) / 1e9
    return float("nan"), float("nan")


def bench(model, prompt=PROMPT, runs=1):
    rows = []
    for _ in range(runs):
        unload(model)                       # cold start, so the load time is visible
        t0 = time.perf_counter()
        r = requests.post(f"{OLLAMA}/api/generate",
                          json={"model": model, "prompt": prompt, "stream": False,
                                "think": False}, timeout=900).json()
        wall = time.perf_counter() - t0
        total_gb, vram_gb = loaded_info(model)
        rows.append({
            "Model": model,
            "Disk size (GB)": next((m["size"] / 1e9 for m in requests.get(f"{OLLAMA}/api/tags").json()["models"]
                                    if m["name"].startswith(model.split(":")[0])), float("nan")),
            "Loaded size (GB)": total_gb,
            "VRAM (GB)": vram_gb,
            "Load time (s)": r.get("load_duration", 0) / 1e9,
            "Response time (s)": wall,
            "Tokens": r.get("eval_count"),
            "Tokens/s": r["eval_count"] / (r["eval_duration"] / 1e9),
            "System RAM used (%)": psutil.virtual_memory().percent,
        })
    return pd.DataFrame(rows)


if __name__ == "__main__":
    import sys
    print(pd.concat([bench(m) for m in sys.argv[1:]]).round(2).to_string(index=False))
````

### Discussion — the model-size trade-off
```
Model size ──► Quality
           ──► Memory
           ──► Latency
           ──► Hardware requirements
```
Local model selection should be based on **actual measurements on the target machine**, not just public benchmark rankings.

**✍️ Your discussion:**

The measurements and the answers pulled in opposite directions. Gemma 3 4B was the cheaper model on every metric. It generated about 210 tokens/s against Qwen's 120 or so, answered the benchmark prompt in about 12.5 seconds instead of 19.5, and used about 2.9 GB of memory versus 5.6 GB.

Quality went the other way. Qwen's answers were better structured and stuck closer to what I actually asked for. Gemma's were correct, but they read as generic and were sometimes inconsistent. So the extra ~7 seconds and ~2.7 GB bought better instruction-following and structure, not better correctness, since both models got the substance right.

Whether that's worth paying for depends on the task. If the output has to follow a specific format or feed into something else, Qwen's reliability is worth the wait. For quick, interactive use or a memory-limited machine like an 8 GB laptop, Gemma is the more practical choice, as long as I'm fine with possibly inconsistent answers. A public leaderboard wouldn't have shown me any of this, because it ranks quality in general and says nothing about speed, memory, or how closely a model follows my particular prompt on my own hardware.

````python
%%writefile LocalAI_Lab/source/chatbot.py
"""Console chatbot with conversation memory (uses /api/chat)."""
import requests

OLLAMA = "http://localhost:11434"
MODEL = "qwen3.5"


def main():
    messages = [{"role": "system", "content": "You are a helpful, concise assistant."}]
    print("Local AI Assistant — type 'exit' to stop.")
    while True:
        q = input("\nYou: ")
        if q.lower() == "exit":
            break
        messages.append({"role": "user", "content": q})
        r = requests.post(f"{OLLAMA}/api/chat",
                          json={"model": MODEL, "messages": messages, "stream": False})
        r.raise_for_status()
        reply = r.json()["message"]
        messages.append({"role": "assistant", "content": reply["content"]})
        print("\nAI:", reply["content"])


if __name__ == "__main__":
    main()
````

````python
%%writefile LocalAI_Lab/source/app.py
"""FSTT Local AI Assistant — Flask + Ollama + local RAG + conversation history."""
import json, os, time
import requests
from flask import Flask, jsonify, render_template, request

import rag

OLLAMA = os.environ.get("OLLAMA_URL", "http://localhost:11434")
MODEL = os.environ.get("MODEL", "qwen3.5")
DOC_DIR = os.path.join(os.path.dirname(__file__), "..", "documents")
MAX_TURNS = 6                      # remembered user/assistant pairs

app = Flask(__name__)
retriever = rag.Retriever(rag.load_documents(DOC_DIR))
history = []                       # [{"role": ..., "content": ...}]  (single-user demo)

SYSTEM = ("You are the FSTT Local AI Assistant. Answer questions about the course using only the "
          "provided CONTEXT. If the answer is not in the context, say: \"Information not available.\" "
          "Be concise.")

# Structured response schema used for "Analyze ..." / "Quiz ..." style questions
STRUCT_SCHEMA = {
    "type": "object",
    "properties": {
        "topic": {"type": "string"},
        "summary": {"type": "string"},
        "key_points": {"type": "array", "items": {"type": "string"}},
        "difficulty": {"type": "string", "enum": ["easy", "medium", "hard"]},
    },
    "required": ["topic", "summary", "key_points", "difficulty"],
}


def call_ollama(messages, structured=False):
    payload = {"model": MODEL, "messages": messages, "stream": False, "think": False}
    if structured:
        payload["format"] = STRUCT_SCHEMA
    r = requests.post(f"{OLLAMA}/api/chat", json=payload, timeout=600)
    r.raise_for_status()
    return r.json()


@app.route("/")
def index():
    return render_template("index.html", model=MODEL, docs=list(retriever.docs))


@app.route("/ask", methods=["POST"])
def ask():
    body = request.get_json(force=True)
    question = (body.get("question") or "").strip()
    structured = bool(body.get("structured"))
    if not question:
        return jsonify(error="Empty question"), 400

    hits, context = retriever.context(question, k=2)
    user_msg = f"CONTEXT:\n{context or '(none)'}\n\nQUESTION: {question}"
    messages = [{"role": "system", "content": SYSTEM}] + history[-2 * MAX_TURNS:] \
               + [{"role": "user", "content": user_msg}]

    t0 = time.perf_counter()
    try:
        out = call_ollama(messages, structured)
    except requests.RequestException as e:
        return jsonify(error=f"Ollama error: {e}"), 502
    elapsed = time.perf_counter() - t0

    answer = out["message"]["content"]
    # store the plain question (not the long context) to keep history small
    history.append({"role": "user", "content": question})
    history.append({"role": "assistant", "content": answer})

    parsed = None
    if structured:
        try:
            parsed = json.loads(answer)
        except json.JSONDecodeError:
            pass

    tokens = out.get("eval_count")
    tps = tokens / (out["eval_duration"] / 1e9) if tokens and out.get("eval_duration") else None
    return jsonify(answer=answer, structured=parsed,
                   sources=[{"doc": n, "score": round(s, 2)} for n, s in hits],
                   seconds=round(elapsed, 2), tokens=tokens,
                   tokens_per_s=round(tps, 1) if tps else None)


@app.route("/reset", methods=["POST"])
def reset():
    history.clear()
    return jsonify(ok=True)


if __name__ == "__main__":
    # Bind to localhost ONLY (see report question Q15 on security)
    app.run(host="127.0.0.1", port=int(os.environ.get("PORT", 5001)), debug=False)
````

## Part 18 — Questions for the Report
Answer each question in your own words (use your measurements from Parts 6 and 16 for Q13–Q14).

**Q1.** What is Local AI?

**✍️ Answer:** Local AI means running the model on my own machine instead of calling a cloud service. The weights are downloaded once and inference happens on my CPU/GPU, so prompts and documents never leave the computer and it works offline.

**Q2.** Difference between a Cloud LLM and a Local LLM?

**✍️ Answer:** A cloud LLM (ChatGPT, Gemini, Claude) runs on the provider's servers. I pay per token, send my data over the internet, and get very large models with no hardware to maintain. A local LLM runs on my own hardware. It is free per request, private and offline-capable, but limited to models that fit in my RAM/VRAM (here 3 to 7 GB models), so it is usually less capable and I'm responsible for setup, updates and speed.

**Q3.** What role does Ollama play?

**✍️ Answer:** Ollama is the runtime that makes local models easy to use. It downloads and stores models (`ollama pull`), loads them into memory with GPU acceleration, manages quantized formats, and exposes a REST API on port 11434 (`/api/generate`, `/api/chat`, `/api/embed`). My Python scripts, the RAG module and the Flask app all talk to the model through it instead of handling weights and inference themselves.

**Q4.** What is model quantization?

**✍️ Answer:** Quantization stores the model's weights with fewer bits (for example 4-bit integers instead of 16-bit floats). This makes the file and the memory footprint 2 to 4 times smaller and speeds up inference, at the cost of a small loss in precision and therefore quality. It is what lets a multi-billion-parameter model fit on a normal laptop or consumer GPU.

**Q5.** Why does model size affect memory consumption?

**✍️ Answer:** Every parameter has to sit in memory while the model runs, so memory is roughly parameters x bytes per weight, plus the KV cache that grows with context length. In my benchmark qwen3.5 took 5.61 GB loaded against 2.88 GB for gemma3, about double, in line with the disk sizes (6.55 GB vs 3.35 GB). If the model doesn't fit in VRAM, layers spill to system RAM and the CPU and generation slows down a lot.

**Q6.** Why can a smaller model sometimes be preferable?

**✍️ Answer:** A smaller model loads faster (6.2 s vs 11.2 s), generates faster (209 vs 122 tokens/s), uses less memory and can run on weak hardware, even without a GPU. For simple tasks such as short answers, summaries or RAG over supplied text, where the context does the heavy lifting, the quality gap is small, so the smaller model is the better trade-off. It also gives lower latency for interactive use and lower power consumption.

**Q7.** Advantages of local inference?

**✍️ Answer:** Privacy (data stays on the machine), no per-token cost, offline operation, no network latency or rate limits, full control over the model version and its behaviour (it won't change or be deprecated under me), and the ability to work with sensitive documents such as student or company data.

**Q8.** Disadvantages of local inference?

**✍️ Answer:** Limited by my hardware: only smaller or quantized models fit, so quality and context length are below the best cloud models. Speed depends on the GPU, large models can be slow or not run at all, and it uses RAM/VRAM and power that other programs need. I also have to install, update and secure everything myself, and there's no built-in scaling for many users.

**Q9.** Explain the architecture of your Python application.

**✍️ Answer:** The app is a Flask server (`app.py`) in front of Ollama. The browser page (`templates/index.html`) sends the question with `fetch` to `/ask`. The server uses the retriever in `rag.py` (TF-IDF keyword search) to pick the top 2 documents from `documents/`, then builds a message list made of a system prompt, the last 6 turns of conversation history and the new question with its CONTEXT. It sends that to Ollama's `/api/chat` (with a JSON schema in `format` when structured output is requested) and returns the answer, the sources, the response time and the tokens/s as JSON, which the page displays. `/reset` clears the history. `chatbot.py` is a console version with memory, and `benchmark.py` measures the models.

**Q10.** Explain the role of the REST API.

**✍️ Answer:** The REST API is the interface between my code and the model. Ollama is an HTTP server, so any language or program can use the model by sending a JSON request (model, prompt/messages, options) and getting a JSON response with the text and counters such as `eval_count` and `eval_duration`. It decouples the client (Flask, notebook, console) from the model runtime, lets me swap models by changing a string, and the same pattern is used between my browser page and Flask (`/ask`, `/reset`).

**Q11.** What is RAG?

**✍️ Answer:** Retrieval-Augmented Generation. Instead of relying only on what the model memorised in training, the system first retrieves the most relevant documents for the question (keyword TF-IDF in my case, or embeddings with nomic-embed-text), puts them in the prompt as context, and tells the model to answer only from it. This gives answers grounded in my own course material, reduces hallucination, allows up-to-date or private knowledge without retraining, and lets the app show its sources. Its weakness is that retrieval quality limits everything: in my test the follow-up question 'how is it different from supervised learning?' returned 'Information not available.' because the retriever only saw the follow-up words.

**Q12.** Difference between an LLM and a multimodal LLM?

**✍️ Answer:** A standard LLM only takes text in and produces text out. A multimodal LLM has additional encoders (for example a vision encoder) that convert images, and in newer models audio or video, into embeddings the language model can process together with the text. In my Part 14 test, gemma3 received the image (base64 through the `images` field) and correctly described the Qatar Airways page, reading the visible text, which a text-only model cannot do.

**Q13.** Compare two local models experimentally.

**✍️ Answer:** I ran the same prompt ('Write a 500-word explanation of reinforcement learning.') on both models with a cold start each time:

| Model | Disk (GB) | Loaded (GB) | Load (s) | Response (s) | Tokens | Tokens/s | Quality /5 |
|---|---|---|---|---|---|---|---|
| qwen3.5 | 6.55 | 5.61 | 11.17 | 19.40 | 735 | 121.79 | 4 |
| gemma3 | 3.35 | 2.88 | 6.24 | 12.48 | 854 | 209.25 | 3 |

Gemma 3 is better on every cost metric: about 1.7x faster, 1.6x quicker to answer, half the memory. Qwen 3.5 gave better structured answers that followed the instructions more closely, which is why I scored it 4 against 3. Both were correct on substance, so the extra cost buys instruction-following and structure, not correctness. This is a single run per model, so the numbers are indicative rather than statistically solid.

**Q14.** Which model would you deploy on a computer with limited RAM? Explain.

**✍️ Answer:** I would deploy gemma3 (4B, quantized). It needs about 2.9 GB loaded versus 5.6 GB for qwen3.5, which leaves room for the OS and the app on an 8 GB machine and avoids swapping or spilling out of VRAM, which would slow everything down. It also loads in about 6 s and answered in 12.5 s, and for RAG over supplied documents its quality was sufficient. If the machine were even smaller I would go to a 1B-class model. I would only choose qwen3.5 if I had the memory and strict format following mattered.

**Q15.** What security risks exist when exposing a local LLM server to a network?

*(Hint: Ollama has no authentication by default; think about `OLLAMA_HOST=0.0.0.0`, unauthorized use of your compute, model/data exposure, prompt injection through documents, and what a reverse proxy with auth/TLS or a firewall would change.)*

**✍️ Answer:** Ollama has no authentication by default, and the lab's first Flask app even runs with `host='0.0.0.0'` and `debug=True`. If I set `OLLAMA_HOST=0.0.0.0` or expose the port, anyone on the network can: use my GPU/CPU for free (denial of service or abuse, since a model can be hammered with huge prompts); list, pull, delete or replace my models via the API; read whatever my app sends and any documents or conversation it exposes; and run my prompts through a model that may have no safety limits. With RAG there is also prompt injection: a malicious instruction hidden inside a document can override the system prompt and leak other context. Flask's debug mode additionally allows remote code execution through the Werkzeug debugger. Mitigations: bind to 127.0.0.1 only (as my final app does), firewall the port, put a reverse proxy (nginx/Caddy) with TLS and authentication/rate limiting in front, use a VPN, disable debug, and treat documents as untrusted input.
